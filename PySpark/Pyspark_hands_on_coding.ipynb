### 5. Initialize SparkSession
Finally, let's initialize a SparkSession to confirm everything is set up correctly.

```python
# Initialize SparkSession
from pyspark.sql import SparkSession

spark = SparkSession.builder.master("local[*]").appName("ColabPySpark").getOrCreate()

print("SparkSession created successfully!")
spark
```

```python
# spark reading the data
dfspark = spark.read.option("header", "true").csv("/content/sample_data/california_housing_test.csv")
```

```python
# check the schema
dfspark.printSchema()
```
