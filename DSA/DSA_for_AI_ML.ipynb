```python
# two sums
def two_sum(nums, target):
  seen = {}
  for i, n in enumerate(nums):
    if target - n in seen:
      return [seen[target - n], i]
    else:
      seen[n] = i

  return []
```

```python
# anagrams
def group_anagrams(strs):
    groups = {}
    for s in strs:
        count = [0] * 26
        for c in s:
            count[ord(c) - ord('a')] += 1
        key = tuple(count)
        groups.setdefault(key, []).append(s)
    return list(groups.values())
```
