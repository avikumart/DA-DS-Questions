```r
ggplot(data = mpg) +
  geom_point(mapping = aes(x = displ, y =hwy)) +
  facet_wrap(~ class, nrow = 2)
```

```r
ggplot(data= mpg) +
  geom_point(mapping = aes(x = displ, y = hwy)) +
  facet_grid(drv ~cyl)
```

```r
ggplot(data = mpg) +
  geom_smooth(mapping = aes(x = displ, y = hwy, color = "orange"))
```

```r
ggplot(data = mpg) +
  geom_smooth(mapping = aes(x = displ, y = hwy, linetype = drv))
```

```r
ggplot(data = mpg) +
  geom_smooth(mapping = aes(x = displ, y=hwy, color=class))
```

```r
# layering multiple plots
ggplot(data = mpg) +
  geom_point(mapping = aes(x = displ, y=cty)) +
  geom_smooth(mapping = aes(x = displ, y=cty))
```

```r
# avoid repetaion of mapping function to the charts
ggplot(data = mpg, mapping = aes(x = displ, y = hwy)) +
  geom_point() +
  geom_smooth()
```

```r
# color different lines of the both the plots
ggplot(data = mpg, mapping = aes(x =displ, y = hwy)) +
  geom_point(mapping = aes(color = class)) +
  geom_smooth()
```

```r
ggplot(data = diamonds) +
  geom_bar(mapping = aes(x = color))
```

```r
# geom bar for the stat summary vizzes
ggplot(data = diamonds) +
  stat_summary(
  mapping = aes(x = cut, y = depth),
  fun.min = min,
  fun.max = max,
  fun = median
  )
```

```r
ggplot(data = diamonds) +
  geom_bar(mapping = aes(x = cut, colour =cut))
```

```r
# filll the color
ggplot(data =diamonds) +
  geom_bar(mapping =aes (x = cut, fill=cut))
```

```r
ggplot(data =diamonds) +
  geom_bar(mapping =aes (x = cut, fill=clarity))
```

```r
ggplot(data = diamonds, mapping =aes(x = cut, fill = clarity)) +
  geom_bar(alpha = 1/5, position ="identity")
```

```r
ggplot(data = diamonds) +
  geom_bar(mapping = aes(x = cut, fill = clarity), position = "fill")
```

```r
ggplot(data = diamonds) +
  geom_bar(mapping = aes(x = cut, fill = clarity), position = "dodge")
```

```r
# geographic data plots in R
library(maps)

us <- map_data("usa")

ggplot(us, aes(long, lat, group=group)) +
  geom_polygon(fill = "white", colour = "black") +
  coord_quickmap()
```

```r
ggplot(mpg, aes(displ, hwy)) +
  geom_point(aes(color = class)) +
  geom_smooth(se = FALSE) +
  labs(title = "Fuel efficiency generally decreases with engine size")
```

```r
ggplot(mpg, aes(displ, hwy)) +
  geom_point(aes(color = class)) +
  geom_smooth(se =TRUE) +
  labs(
  title = "Fuel efficiency generally decreases with engine size",
  subtitle = "Two seaters are an exception because of their light weight",
  caption = "Data from fueleconomy.gov"
  )
```
