```r
# install libs
library(tidyverse)
library(readr)
library(dplyr)
library(rvest)
library(xml2)
```

```r
# take covid url for the web scraping
covid.url <- 'https://www.worldometers.info/coronavirus/#countries'
```

```r
# get the xpath
xpath.country <- '//*[@id="main_table_countries_today"]'
```

```r
# scrap the data using above details
covid.dt <- covid.url %>%
  read_html() %>%
  html_nodes(xpath=xpath.country) %>%
  html_table(header=TRUE) %>%
  .[[1]] %>%
  filter(!is.na(`#`))
head(covid.dt, 10)
```

```r
# good reads url
goodreads.url <- "https://www.goodreads.com/list/show/191616"
```

```r
# xpath of the goodreads
xpath.goodreads <- '//*[@id="bodycontainer"]'
```

```r
# scrap the data using the rvest web scraping
good.data <- goodreads.url %>%
  read_html() %>%
  html_nodes(xpath=xpath.goodreads) %>%
  html_table()
good.data
```
