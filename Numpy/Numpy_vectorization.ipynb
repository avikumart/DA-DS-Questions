```python
# loop based indexing instea conditional loops
import numpy as np

%timeit
data = np.random.randn(10000000)
result = []
for x in data:
  if x >0:
    result.append(x*2)
  else:
    result.append(x)
result = np.array(result)
result
```

```python
# boolean indexing method
%timeit
data = np.random.randn(10000000)
result = data[data>0]*2
result
```

```python
# broadcasting for the implicit loops
matrix = np.random.rand(1000,500)
row_means = matrix.mean(axis=1)
centered = np.zeros_like(matrix)
for i in range(matrix.shape[0]):
  centered[i] = matrix[i] - row_means[i]
centered.shape
```

```python
# faster method of the broadcasting
matrix = np.random.rand(1000,500)
row_means = matrix.mean(axis=1)
centered = matrix - row_means[:,np.newaxis]
centered.shape
```

```python
matrix = np.random.rand(1000,500)
row_means = np.mean(matrix,axis=1)
centered = matrix - row_means[:,np.newaxis]
centered.shape
```

```python
# use of the np.where in the vector operations
temps = np.random.uniform(-10,10,100000)
classifs = np.select([temps <0, temps<20, temps >=20],
                     ["freezing","cool","warm"],
                     default="unknown"
                     )
classifs
```

```python
# use np.vectorization for the vectorization of the data
def transformation(x):
  if x < 0:
    return np.sqrt(abs(x))
  else:
    return x**2
```

```python
vec_trans = np.vectorize(transformation)
data = np.random.randn(100002)
results = vec_trans(data)
results
```

```python
# apply along the axis function
data = np.random.rand(1000,50)

def custom_stat(row):
  return (np.max(row) - np.min(row)) / np.median(row)

row_stats = np.apply_along_axis(custom_stat, axis=0, arr=data)
row_stats
```
